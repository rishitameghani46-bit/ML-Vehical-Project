# 🚗 Vehicle Insurance Fraud Detection — Improved ML Model

### Goal
Build a clean, reproducible classification pipeline for the `fraud reported` target.

This version improves the original notebook by:
- cleaning invalid values such as `*`
- converting numeric-looking object columns correctly
- extracting useful information from `claim_date`
- using a proper train/test split
- preprocessing numeric and categorical features inside a single pipeline
- comparing several tree-based models
- tuning the strongest candidate
- reporting **Accuracy, Precision, Recall, F1, ROC-AUC and Confusion Matrix**
- saving the final complete pipeline with `joblib`

> **Important:** An exact 80–85% test accuracy cannot honestly be guaranteed before running the notebook on the actual CSV. The notebook is designed to target that range without artificially changing the test labels or leaking the target into the features.


```python
# 1. Import libraries

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.model_selection import train_test_split, StratifiedKFold, RandomizedSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    HistGradientBoostingClassifier,
    GradientBoostingClassifier
)
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

print("Libraries loaded successfully.")
```

## 2. Load the dataset

```python
# Keep the CSV in the same folder as this notebook.
DATA_PATH = "insurance_fraud_data.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)
display(df.head())
```

```python
# 3. Basic inspection

print("Columns:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes)

print("\nMissing values:")
display(df.isnull().sum().sort_values(ascending=False).head(15))

print("\nDuplicate rows:", df.duplicated().sum())
```

## 4. Data cleaning

The original dataset contains a few values that need special handling:
- `fraud reported` has missing values
- `claim_date` contains `*`
- `injury_claim` and `age_of_vehicle` can be stored as object/string values

The target is kept strictly separate from preprocessing so that the model does not see the answer during training.


```python
# 4A. Clean target

df = df.copy()

df["fraud reported"] = (
    df["fraud reported"]
    .astype("string")
    .str.strip()
    .str.upper()
)

# Remove rows where the target itself is missing
df = df.dropna(subset=["fraud reported"]).copy()

# Convert target: N = 0, Y = 1
df["fraud reported"] = df["fraud reported"].map({"N": 0, "Y": 1})

# Remove any unexpected target values
df = df.dropna(subset=["fraud reported"]).copy()
df["fraud reported"] = df["fraud reported"].astype(int)

print("Cleaned dataset shape:", df.shape)
print("\nTarget distribution:")
print(df["fraud reported"].value_counts())
print("\nTarget percentage:")
print((df["fraud reported"].value_counts(normalize=True) * 100).round(2))
```

```python
# 4B. Convert known numeric columns safely

numeric_like_columns = [
    "age_of_driver",
    "safety_rating",
    "annual_income",
    "high_education",
    "address_change",
    "zip_code",
    "past_num_of_claims",
    "liab_prct",
    "police_report",
    "vehicle_price",
    "total_claim",
    "injury_claim",
    "policy deductible",
    "annual premium",
    "days open",
    "form defects"
]

for col in numeric_like_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

print("Numeric conversion completed.")
```

## 5. Feature engineering

```python
# Create useful features from claim date and financial values

def add_features(data):
    data = data.copy()

    # Date features
    if "claim_date" in data.columns:
        parsed_date = pd.to_datetime(
            data["claim_date"].replace("*", np.nan),
            errors="coerce"
        )

        data["claim_year"] = parsed_date.dt.year
        data["claim_month"] = parsed_date.dt.month
        data["claim_day"] = parsed_date.dt.day
        data["claim_week"] = parsed_date.dt.isocalendar().week.astype("float")
        data["claim_is_weekend"] = parsed_date.dt.dayofweek.isin([5, 6]).astype(float)

        # Raw date string is not needed after extracting its information.
        data = data.drop(columns=["claim_date"])

    # Useful ratios
    if {"injury_claim", "total_claim"}.issubset(data.columns):
        data["injury_claim_ratio"] = (
            data["injury_claim"] / data["total_claim"].replace(0, np.nan)
        )

    if {"total_claim", "vehicle_price"}.issubset(data.columns):
        data["claim_vehicle_ratio"] = (
            data["total_claim"] / data["vehicle_price"].replace(0, np.nan)
        )

    if {"annual_premium", "vehicle_price"}.issubset(data.columns):
        data["premium_vehicle_ratio"] = (
            data["annual premium"] / data["vehicle_price"].replace(0, np.nan)
        )

    if {"total_claim", "annual_income"}.issubset(data.columns):
        data["claim_income_ratio"] = (
            data["total_claim"] / data["annual_income"].replace(0, np.nan)
        )

    return data


df_model = add_features(df)

print("Shape after feature engineering:", df_model.shape)
display(df_model.head())
```

```python
# 6. Remove identifiers / columns with little predictive meaning

# claim_number is an identifier, not a meaningful behavioural feature.
# We also remove very high-cardinality zip code because treating each ZIP as a
# separate category can overfit on a small dataset.

drop_columns = [
    "claim_number",
    "zip_code"
]

# vehicle_color is retained because it is a real categorical vehicle feature.
# Gender/marital status are also retained because the goal here is reproducing
# the dataset's predictive task; remove them if your project requires fairness
# restrictions on those attributes.

drop_columns = [c for c in drop_columns if c in df_model.columns]

X = df_model.drop(columns=["fraud reported"] + drop_columns)
y = df_model["fraud reported"]

print("Final X shape:", X.shape)
print("Final y shape:", y.shape)
print("\nFeatures:")
print(X.columns.tolist())
```

## 7. Train/Test split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows :", len(X_test))
print("Training fraud rate:", round(y_train.mean() * 100, 2), "%")
print("Testing fraud rate :", round(y_test.mean() * 100, 2), "%")
```

## 8. Preprocessing pipeline

```python
numeric_features = X_train.select_dtypes(include=["number"]).columns.tolist()
categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool", "string"]
).columns.tolist()

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

print("Numeric features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
```

## 9. Compare stronger models

```python
models = {
    "Random Forest": RandomForestClassifier(
        n_estimators=350,
        max_depth=12,
        min_samples_leaf=2,
        class_weight=None,
        random_state=42,
        n_jobs=-1
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=350,
        max_depth=14,
        min_samples_leaf=2,
        class_weight=None,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=180,
        learning_rate=0.05,
        max_depth=3,
        min_samples_leaf=5,
        random_state=42
    )
}

results = []
fitted_pipelines = {}

for name, model in models.items():

    pipe = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    pipe.fit(X_train, y_train)

    pred = pipe.predict(X_test)
    prob = pipe.predict_proba(X_test)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred, zero_division=0),
        "Recall": recall_score(y_test, pred, zero_division=0),
        "F1": f1_score(y_test, pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, prob)
    })

    fitted_pipelines[name] = pipe

results_df = pd.DataFrame(results).sort_values("Accuracy", ascending=False)
display(results_df.round(4))
```

## 10. Hyperparameter tuning

The original notebook selected a decision tree even though its test accuracy was substantially below the requested range. Here we tune a tree ensemble instead, while evaluating the model on data that was not used for fitting.


```python
# Tune Extra Trees because it is a strong baseline for mixed tabular data.

extra_pipe = Pipeline([
    ("preprocessor", preprocessor),
    ("model", ExtraTreesClassifier(
        random_state=42,
        n_jobs=-1
    ))
])

param_distributions = {
    "model__n_estimators": [250, 350, 500],
    "model__max_depth": [8, 10, 12, 14, 16, None],
    "model__min_samples_split": [2, 5, 10],
    "model__min_samples_leaf": [1, 2, 4],
    "model__max_features": ["sqrt", "log2", None],
    "model__class_weight": [None, "balanced"]
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

search = RandomizedSearchCV(
    estimator=extra_pipe,
    param_distributions=param_distributions,
    n_iter=18,
    scoring="accuracy",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

search.fit(X_train, y_train)

print("Best CV accuracy:", round(search.best_score_, 4))
print("\nBest parameters:")
print(search.best_params_)
```

## 11. Final model evaluation

```python
final_model = search.best_estimator_

final_pred = final_model.predict(X_test)
final_prob = final_model.predict_proba(X_test)[:, 1]

final_accuracy = accuracy_score(y_test, final_pred)
final_precision = precision_score(y_test, final_pred, zero_division=0)
final_recall = recall_score(y_test, final_pred, zero_division=0)
final_f1 = f1_score(y_test, final_pred, zero_division=0)
final_auc = roc_auc_score(y_test, final_prob)

print("========== FINAL MODEL ==========")
print(f"Accuracy : {final_accuracy:.4f}  ({final_accuracy * 100:.2f}%)")
print(f"Precision: {final_precision:.4f}")
print(f"Recall   : {final_recall:.4f}")
print(f"F1-score : {final_f1:.4f}")
print(f"ROC-AUC  : {final_auc:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    final_pred,
    target_names=["Not Fraud", "Fraud"],
    zero_division=0
))
```

```python
# 12. Confusion matrix

cm = confusion_matrix(y_test, final_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Not Fraud", "Fraud"],
    yticklabels=["Not Fraud", "Fraud"]
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Final Model - Confusion Matrix")
plt.show()
```

## 13. Accuracy target check

This cell does not fake or force the accuracy. It simply reports whether the untouched test set falls into the requested 80–85% range.

If the result is below 80%, do **not** change `y_test`, train on the test set, or copy predictions into the labels. Instead, inspect the feature engineering/model results and tune using only the training data.


```python
print(f"Final test accuracy: {final_accuracy * 100:.2f}%")

if 0.80 <= final_accuracy <= 0.85:
    print("✅ Accuracy is inside the requested 80–85% range.")
elif final_accuracy < 0.80:
    print("⚠️ Accuracy is below 80% on this split. The dataset may not support the target range with these features.")
else:
    print("ℹ️ Accuracy is above 85%. This is not a problem; it simply exceeds the requested minimum range.")
```

## 14. Save the complete model

```python
# Save the entire preprocessing + model pipeline.
# This means the API/application can send raw rows without manually encoding them.

MODEL_PATH = "vehicle_fraud_final_pipeline.pkl"

joblib.dump(final_model, MODEL_PATH)

print(f"Saved: {MODEL_PATH}")
```

## 15. Load the saved model and test one real row

```python
loaded_model = joblib.load(MODEL_PATH)

# Take one real feature row from the test set.
sample = X_test.iloc[[0]].copy()

sample_prediction = loaded_model.predict(sample)[0]
sample_probability = loaded_model.predict_proba(sample)[0, 1]

print("Prediction:", "Fraud" if sample_prediction == 1 else "Not Fraud")
print(f"Fraud probability: {sample_probability * 100:.2f}%")
```

# ✅ Final project flow

**Dataset → Cleaning → Feature Engineering → Train/Test Split → Preprocessing Pipeline → Model Comparison → Hyperparameter Tuning → Final Evaluation → Save Pipeline → Prediction**

### What was fixed compared with the previous notebook

1. Categorical values are no longer blindly converted using `pd.factorize`.
2. The target is mapped explicitly: `N = 0`, `Y = 1`.
3. Invalid dates such as `*` are handled safely.
4. Useful date features are extracted.
5. `injury_claim` and other numeric-looking columns are converted correctly.
6. Preprocessing is fitted only through the training pipeline.
7. Cross-validation is performed on a complete pipeline, so categorical values do not cause the previous `"could not convert string to float"` error.
8. The final `.pkl` contains both preprocessing and the trained model.
9. The notebook evaluates the untouched test set honestly.

> **Note:** No ML notebook can promise 100% real-world accuracy. The correct goal is a strong, validated model with no data leakage and reliable performance on unseen data.
